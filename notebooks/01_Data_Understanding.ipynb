```python
import re
import string

def remove_punctuation(text):
    # Replace every punctuation character with a space
    text = re.sub(f"[{re.escape(string.punctuation)}]", " ", text)

    # Replace multiple spaces with a single space
    text = re.sub(r"\s+", " ", text)

    return text.strip()
```

```python
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].apply(remove_punctuation)
```

```python
idx = 0

print("BEFORE:\n")
print(resume_df.loc[idx, "Resume_str"][:500])

print("\n" + "=" * 80 + "\n")

print("AFTER:\n")
print(resume_df.loc[idx, "Cleaned_Resume"][:500])
```

```python
# Start fresh
resume_df["Cleaned_Resume"] = resume_df["Resume_str"]
```

```python
# 1. Lowercase
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].str.lower()

# 2. Remove URLs
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].apply(remove_urls)

# 3. Remove emails
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].apply(remove_emails)

# 4. Remove phone numbers
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].apply(remove_phone_numbers)

# 5. Remove HTML tags
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].apply(remove_html_tags)

# 6. Remove punctuation (NEW FUNCTION)
resume_df["Cleaned_Resume"] = resume_df["Cleaned_Resume"].apply(remove_punctuation)
```

```python
idx = 0

print("BEFORE:\n")
print(resume_df.loc[idx, "Resume_str"][:500])

print("\n" + "=" * 80 + "\n")

print("AFTER:\n")
print(resume_df.loc[idx, "Cleaned_Resume"][:500])
```

```python
def remove_extra_whitespace(text):
    return re.sub(r"\s+", " ", text).strip()
```

```python
resume_df["Cleaned_Resume"] = (
    resume_df["Cleaned_Resume"]
    .apply(remove_extra_whitespace)
)
```

```python
import nltk

nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
from nltk.tokenize import word_tokenize
```

```python
resume_df["Tokens"] = resume_df["Cleaned_Resume"].apply(word_tokenize)
```

```python
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
```

```python
stop_words = set(stopwords.words("english"))
```

```python
resume_df["Tokens"] = resume_df["Tokens"].apply(
    lambda tokens: [word for word in tokens if word not in stop_words]
)
```

```python
idx = 0

print("Tokens after stopword removal:\n")
print(resume_df.loc[idx, "Tokens"][:40])

print("\nNumber of tokens:")
print(len(resume_df.loc[idx, "Tokens"]))
```

```python
import nltk
nltk.download("wordnet")
nltk.download("omw-1.4")
```

```python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()
```

```python
resume_df["Tokens"] = resume_df["Tokens"].apply(
    lambda tokens: [lemmatizer.lemmatize(word) for word in tokens]
)
```

```python
idx = 0

print("First 40 lemmatized tokens:\n")
print(resume_df.loc[idx, "Tokens"][:40])

print("\nTotal Tokens:")
print(len(resume_df.loc[idx, "Tokens"]))
```

```python
resume_df["Processed_Resume"] = resume_df["Tokens"].apply(
    lambda tokens: " ".join(tokens)
)
```

```python
idx = 0

print("Processed Resume:\n")
print(resume_df.loc[idx, "Processed_Resume"][:300])

print("\nType:")
print(type(resume_df.loc[idx, "Processed_Resume"]))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    min_df=2,
    max_df=0.95
)
```

```python
X = tfidf.fit_transform(resume_df["Processed_Resume"])
```

```python
print("Shape of TF-IDF Matrix:")
print(X.shape)

print("\nType:")
print(type(X))
```

```python
print("Unique Categories:")
print(sorted(resume_df["Category"].unique()))

print("\nTotal Categories:")
print(resume_df["Category"].nunique())
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
label_encoder = LabelEncoder()
```

```python
y = label_encoder.fit_transform(resume_df["Category"])
```

```python
print("Shape of y:")
print(y.shape)

print("\nFirst 10 Encoded Labels:")
print(y[:10])

print("\nNumber of Classes:")
print(len(label_encoder.classes_))

print("\nClasses:")
print(label_encoder.classes_)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)
```

```python
print("Training Feature Matrix:", X_train.shape)
print("Testing Feature Matrix :", X_test.shape)

print()

print("Training Labels:", y_train.shape)
print("Testing Labels :", y_test.shape)
```

# Train the First Model (Multinomial Naive Bayes)

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
nb_model = MultinomialNB()
```

```python
nb_model.fit(X_train, y_train)
```

```python
y_pred = nb_model.predict(X_test)
```

```python
print("Number of Predictions:", len(y_pred))
print("First 20 Predictions:")
print(y_pred[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.4f}")
print(f"Accuracy (%): {accuracy * 100:.2f}%")
```

```python
print(classification_report(
    y_test,
    y_pred,
    target_names=label_encoder.classes_,
    zero_division=0
))
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```
